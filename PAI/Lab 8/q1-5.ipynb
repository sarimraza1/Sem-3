```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data.csv")
```

QUESTION 1

```python
data = df[df["scenario"] == "01_StudyVsScore"]

sns.scatterplot(data=data, x="study_hours", y="exam_score",hue="prep_course", style="prep_course")
plt.title("Study Hours vs. Exam Score by Prep Course Status")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)

plt.show()
```

QUESTION 2

```python
data = df[df["scenario"] == "02_MonthlySales"]

tech = data[data["school_type"] == "TechStore"]
apparel = data[data["school_type"] == "ApparelCo"]
plt.plot(tech["study_hours"], tech["exam_score"], marker="o", label="TechStore")
plt.plot(apparel["study_hours"], apparel["exam_score"], marker="o", label="ApparelCo")
plt.title("Monthly Revenue Trends (Jan - Jun 2023)")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")

plt.legend()
plt.show()
```

QUESTION 3

```python
data = df[df["scenario"] == "03_SalaryDiff"]

sns.boxplot(data=data, x="prep_course", y="exam_score",hue="prep_course", order=["Entry", "Mid", "Senior"],palette="Set2", legend=False)
plt.title("Salary Spread Across Experience Levels")
plt.xlabel("Experience Level")
plt.ylabel("Salary ($)")

plt.show()
```

QUESTION 4

```python
data = df[df["scenario"] == "04_CustAgeDist"]
sns.histplot(data=data, x="prep_course", hue="study_hours",bins=5, kde=True)
plt.title("Customer Age Distribution by Gender")
plt.xlabel("Age")
plt.ylabel("Customer Count")

plt.show()
```

QUESTION 5

```python
data = df[df["scenario"] == "05_AdVsRevenue"].copy()

data["study_hours"] = pd.to_numeric(data["study_hours"])
data["prep_course"] = pd.to_numeric(data["prep_course"])

sns.regplot(data=data, x="study_hours", y="prep_course",marker="+", color="darkgreen")
plt.title("Impact of Advertising Spend on Total Revenue")
plt.xlabel("Ad Spend ($)")
plt.ylabel("Revenue ($)")
plt.grid(True)

plt.show()
```
